## Colour palettes in seaborn and matplotlib

Seaborn being built on matplotlib makes use of all the matplotlib colour palettes - for their names, have a look here: https://matplotlib.org/stable/users/explain/colors/colormaps.html

Moreover, Seaborn also accepts e.g. "deep", "Set2", "viridis", "rocket", "crest", "vlag".

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd
```

### Sequential palettes

Perceptionally uniform: colour difference mapped to difference in value (extreme values might be difficult to plot).

```python
# Seaborn color palettes: "mako", "flare", "crest"; can be reversed by adding "_r"
sns.color_palette("flare", as_cmap=True)
```

```python
# Seaborn color palettes: "mako", "flare", "crest"; can be reversed by adding "_r"
sns.color_palette("rocket_r", as_cmap=True)
```

```python
# Seaborn cubehelix palettes
sns.cubehelix_palette(as_cmap=True)
```

```python
# Matplotlib palettes: "viridis", "cividis", "magma"
sns.color_palette("viridis", as_cmap=True)
```

```python
# Custom cubehelix palette: 
# start = starting position in color wheel (0-3) ...see here: https://color-space.io/cubehelix 
# rot = rotation of colour wheel = change of hues
# dark and light: darkest and lightest colours on the palette
sns.cubehelix_palette(start=207, rot=1.76, dark=0.35, light=0.5, n_colors=10)
```

### Qualitative color palettes

Qualitative palettes are well-suited to representing categorical data because most of their variation is in the hue component.

```python
# Deafult palette in seaborn:
sns.color_palette()
```

```python
# many other palettes available! Try different ones!
sns.color_palette("pastel")
```

```python
# circular pattern: evenly spaced hues in a circle, constant brightness and saturation.
# Perception by human eye: different intensity for different colours (RGB issue!)
sns.color_palette("hls", 10)
```

```python
# Create a sequential palette with 6 colors
sns.color_palette("Blues", 5)
```

### Diverging palettes
If extreme values are interesting and the middle is less so (e.g. correlation): Similar to sequential, but with two colours.

```python
# Seaborn palettes: e.g. "vlag", "icefire"
sns.color_palette("icefire", as_cmap=True)
```

```python
# Matplotlib palettes: e.g. "Spectral", "coolwarm"
sns.color_palette("coolwarm", as_cmap=True)
```

```python
# Custom diverging palette, using the hus1 system: define two angles on colour wheel and saturation (s)
sns.diverging_palette(15, 250, s=90, as_cmap=True)
```

### Apply colour palettes to plots:

Look at the following plots and apply the (your?) ideal colour palette.

```python
# Data created here...

rng = np.random.default_rng(42)

# Data 1
t = np.arange(0, 20)
groups = ["A", "B", "C", "D"]
df_cats = pd.DataFrame(
    [
        {"time": ti, "group": g, "value": 2 * k + 0.4 * k * ti + rng.normal(0, 1.2)}
        for k, g in enumerate(groups, start=1)
        for ti in t
        for _ in range(3)  # 3 replicates -> confidence band
    ]
)

# Data 2
n = 200
df_scatter = pd.DataFrame({"x": rng.normal(size=n), "y": rng.normal(size=n)})
df_scatter["temperature"] = (
    20 + 8 * df_scatter["x"] + 5 * df_scatter["y"] + rng.normal(0, 2, n)
)

# Data 3: 
base = rng.normal(size=(300, 3))
features = pd.DataFrame(
    {
        "f1": base[:, 0],
        "f2": base[:, 0] * 0.8 + rng.normal(0, 0.5, 300),
        "f3": -base[:, 0] * 0.6 + rng.normal(0, 0.7, 300),
        "f4": base[:, 1],
        "f5": base[:, 1] * 0.7 + rng.normal(0, 0.6, 300),
        "f6": base[:, 2],
        "f7": -base[:, 2] * 0.9 + rng.normal(0, 0.4, 300),
        "f8": rng.normal(size=300),
    }
)
corr = features.corr()
```

And here come the plots:

```python
# Plot 1
sc = sns.scatterplot(
    data=df_cats, x="time", y="value", hue="group",
    palette=["steelblue"] * 4,
)

plt.show()
```

```python
# Plot 2
sc = sns.scatterplot(
    data=df_scatter, x="x", y="y", hue="temperature",
    palette=["blue"] * 4, #Creates a list of four different values, not multiplication
    legend=False
)
plt.show()
```

```python
from matplotlib.colors import ListedColormap
sns.heatmap(
    corr, vmin=-1, vmax=1, center=0,
    annot=True, fmt=".2f", square=True, linewidths=0.5,
    cbar_kws={"label": "Pearson r", "shrink": 0.8},
    cmap="vlag"
)
plt.show()
```
